- [ ] Color-code output using the color scheme specified below:

    | **Variable Type** | Color |
    | :---------------- | :---- |
    | $(a, b)$ |  `#00a6ff` |
    | $\gcd{(a, b)}$ | `#28d728` |
    | Bézout's Coefficients, $x$ and $y$ | `#ff00c3` |

```
from __future__ import annotations

from kuramoto.cli.style import (
    PRESETS,
    cli_divider,
    cli_echo,
    cli_section,
    multi_gradient,
    style_coprime_coeff,
    style_coprime_gcd,
    style_coprime_term,
)
from kuramoto.solvers.coprime import extended_gcd

def trace_extended_gcd(a: int, b: int) -> tuple[int, int, int]:
    """
    Traces recursive calls of extended_gcd and prints Bézout identity with:
    - Signature Electric Blue -> Purple -> Magenta gradient on step labels
    - Color-coded terms (#00a6ff), gcd (#28d728), and coefficients (#ff00c3)
    """
    def _traced(a_val: int, b_val: int, depth: int) -> tuple[int, int, int]:
        indent = "  " * depth
        label_call = multi_gradient(f"Call depth {depth}:", PRESETS["electric"])
        cli_echo(
            f"{indent}{label_call} _ext_gcd(a={style_coprime_term(a_val)}, b={style_coprime_term(b_val)})"
        )

        if a_val == 0:
            label_base = multi_gradient("Base case reached:", PRESETS["electric"])
            cli_echo(
                f"{indent}  ↳ {label_base} gcd={style_coprime_gcd(b_val)}, x={style_coprime_coeff(0)}, y={style_coprime_coeff(1)}"
            )
            return b_val, 0, 1

        r = b_val % a_val
        q = b_val // a_val
        g, a1, b1 = _traced(r, a_val, depth + 1)
        x = b1 - q * a1
        y = a1

        label_unwind = multi_gradient("Unwinding:", PRESETS["electric"])
        cli_echo(
            f"{indent}  ↳ {label_unwind} ({style_coprime_coeff(x)}){style_coprime_term(a_val)} + ({style_coprime_coeff(y)}){style_coprime_term(b_val)} = {style_coprime_gcd(g)}"
        )
        return g, x, y

    cli_section(f"Recursive Extended GCD Trace: ({a}, {b})")
    g, x, y = _traced(a, b, depth=0)
    cli_divider()

    # Print raw tuple and formatted verification line
    print((g, x, y))
    cli_echo(
        f"Bezout's Identity for {style_coprime_term(a)} and {style_coprime_term(b)}: "
        f"({style_coprime_coeff(x)}){style_coprime_term(a)} + ({style_coprime_coeff(y)}){style_coprime_term(b)} = {style_coprime_gcd(g)} ✔️"
    )
    return g, x, y
```

```
# Bezout's Identity for 7 and 11
a1 = 7
b1 = 11
gcd_a1_b1 = trace_extended_gcd(a1, b1)
```
