```
from __future__ import annotations

from kuramoto.solvers.coprime import extended_gcd
```

```
# Color helper functions with corrected ANSI TrueColor escape sequences
def c_term(val: int | str) -> str:
    """Color (a, b) terms in cyan (#00a6ff)."""
    return f"\033[38;2;0;166;255m{val}\033[0m"


def c_gcd(val: int | str) -> str:
    """Color gcd values in green (#28d728)."""
    return f"\033[38;2;40;215;40m{val}\033[0m"


def c_coeff(val: int | str) -> str:
    """Color Bezout coefficients (x, y) in magenta (#ff00c3)."""
    return f"\033[38;2;255;0;195m{val}\033[0m"

def trace_extended_gcd(a: int, b: int) -> tuple[int, int, int]:
    """
    Traces recursive calls and returns (gcd, x, y) with color-coded steps.
    """
    call_depth = 0

    def _traced_ext_gcd(a_val: int, b_val: int, depth: int) -> tuple[int, int, int]:
        indent = "  " * depth
        cli_coprime(
            f"{indent}Call depth {depth}: _ext_gcd(a={c_term(a_val)}, b={c_term(b_val)})"
        )

        if a_val == 0:
            click.echo(
                f"{indent}  ↳ Base case reached: gcd={c_gcd(b_val)}, x={c_coeff(0)}, y={c_coeff(1)}"
            )
            return b_val, 0, 1

        r = b_val % a_val
        q = b_val // a_val
        g, a1, b1 = _traced_ext_gcd(r, a_val, depth + 1)
        x = b1 - q * a1
        y = a1

        click.echo(
            f"{indent}  ↳ Unwinding: {c_term(a_val)}·({c_coeff(x)}) + {c_term(b_val)}·({c_coeff(y)}) = {c_gcd(g)}"
        )
        return g, x, y

    cli_section(f"Recursive Extended GCD Trace: ({a}, {b})")
    g, x, y = _traced_ext_gcd(a, b, depth=0)
    cli_divider()
    cli_success(
        f"Verified Identity: {c_term(a)}·({c_coeff(x)}) + {c_term(b)}·({c_coeff(y)}) = {c_gcd(g)} ✔️"
    )
    return g, x, y

def format_bezout(a: int, b: int) -> tuple[int, int, int]:
    """Compute and print colored Bezout identity matching the script's format."""
    gcd_val, x, y = extended_gcd(a, b)

    # 1. Print raw tuple result
    print(gcd_val, x, y)

    # 2. Print colored identity matching the format: (x)a + (y)b = gcd ✔️
    formatted_str = (
        f"Bezout's Identity for {c_term(a)} and {c_term(b)}: "
        f"({c_coeff(x)}){c_term(a)} + ({c_coeff(y)}){c_term(b)} = {c_gcd(gcd_val)} ✔️"
    )
    cli_echo(formatted_str)
    return gcd_val, x, y


# --- Execute for Part a/d: (7, 11) ---
g1, x1, y1 = trace_extended_gcd(7, 11)

# --- Execute for Part c/d: (6, 9) ---
g2, x2, y2 = trace_extended_gcd(6, 9)
```
